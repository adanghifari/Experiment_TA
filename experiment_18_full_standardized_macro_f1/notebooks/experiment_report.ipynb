# Experiment Report

Notebook ini disiapkan untuk membaca hasil retrain dari folder `results/`. Jalankan setelah `run_experiment.ps1` selesai.

```python
from pathlib import Path
import json
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
RESULTS = ROOT / 'results'
PROTOCOL = ROOT / 'protocol.json'

protocol = json.loads(PROTOCOL.read_text())
protocol
```

```python
def load_json(path):
    path = Path(path)
    return json.loads(path.read_text()) if path.exists() else None

front = load_json(RESULTS / 'front' / 'test_metrics.json')
side = load_json(RESULTS / 'side' / 'test_metrics.json')
fusion = load_json(RESULTS / 'fusion' / 'comparison.json')

rows = []
for name, data in [('Front model', front), ('Side model', side)]:
    if data:
        rows.append({
            'method': name,
            'accuracy': data.get('accuracy'),
            'precision_macro': data.get('precision_macro'),
            'recall_macro': data.get('recall_macro'),
            'f1_macro': data.get('f1_macro'),
        })
if fusion:
    for key, label in [('average_fusion', 'Average fusion'), ('adaptive_fusion', 'Adaptive fusion')]:
        data = fusion.get(key, {})
        rows.append({
            'method': label,
            'accuracy': data.get('accuracy'),
            'precision_macro': data.get('precision_macro'),
            'recall_macro': data.get('recall_macro'),
            'f1_macro': data.get('f1_macro'),
        })

pd.DataFrame(rows)
```

```python
# Confusion matrices
for name, data in [('front', front), ('side', side)]:
    if data:
        print(name, data.get('confusion_matrix'))
if fusion:
    print('average_fusion', fusion.get('average_fusion', {}).get('confusion_matrix'))
    print('adaptive_fusion', fusion.get('adaptive_fusion', {}).get('confusion_matrix'))
```
